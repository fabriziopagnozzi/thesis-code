## Examples

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
rng = np.random.default_rng(42)
n_docs = 200


df = pd.DataFrame(
    {
        'relevance': rng.uniform(0, 1, n_docs),
        'diversity': rng.uniform(0, 1, n_docs),
        'category': rng.choice(['sports', 'tech', 'science', 'politics'], n_docs),
    }
)

print(df.head(), '\n' * 2)
print(df.describe())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

sns.scatterplot(data=df, x='relevance', y='diversity', hue='category', ax=axes[0])
axes[0].set_title('Relevance vs Diversity')

sns.boxplot(data=df, x='category', y='relevance', ax=axes[1])
axes[1].set_title('Relevance by Category')

plt.tight_layout()
plt.show()
```

```python
# --- Heatmap (correlation matrix) ---
fig, ax = plt.subplots(figsize=(3, 3))

corr = df[['relevance', 'diversity']].corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', ax=ax)
ax.set_title('Correlation Matrix')

plt.tight_layout()
plt.show()
```

```python
# --- Bar plot + Violin ---
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

sns.barplot(data=df, x='category', y='relevance', ax=axes[0])
axes[0].set_title('Mean Relevance by Category')

sns.violinplot(data=df, x='category', y='diversity', ax=axes[1])
axes[1].set_title('Diversity Distribution by Category')

plt.tight_layout()
plt.show()
```

```python
# --- Histogram + KDE ---
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(data=df, x='relevance', hue='category', kde=True, ax=axes[0])
axes[0].set_title('Relevance Distribution')

sns.kdeplot(data=df, x='relevance', y='diversity', hue='category', ax=axes[1])
axes[1].set_title('Relevance vs Diversity (Density)')

plt.tight_layout()
plt.show()
```
